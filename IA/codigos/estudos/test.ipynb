```python
import json

import soundfile as sf
import librosa
import io
import os
import matplotlib.pyplot as plt

import numpy as np
import librosa.display
```

```python
audio1, sr1 = librosa.load("/mnt/hd/audios_backup_test/VITRECTOMIA.wav", sr=16000)
audio2, sr2 = librosa.load("/mnt/hd/audios_backup_test/TROPONINA.wav", sr=16000)

plt.figure(figsize=(10, 4))
librosa.display.waveshow(audio1, sr=sr1)
plt.title("Forma de Onda")
plt.xlabel("Tempo (s)")
plt.ylabel("Amplitude")
plt.show()

```

```python
plt.figure(figsize=(10, 4))
librosa.display.waveshow(audio2, sr=sr2)
plt.title("Forma de Onda")
plt.xlabel("Tempo (s)")
plt.ylabel("Amplitude")
plt.show()
```

```python

espectrograma = np.abs(librosa.stft(audio1))

# Exibir o espectrograma
plt.figure(figsize=(10, 4))
librosa.display.specshow(librosa.amplitude_to_db(espectrograma, ref=np.max), sr=sr1, x_axis="time", y_axis="log")
plt.title("Espectrograma")
plt.colorbar(format="%+2.0f dB")
plt.show()
```

```python
espectrograma = np.abs(librosa.stft(audio2))

# Exibir o espectrograma
plt.figure(figsize=(10, 4))
librosa.display.specshow(librosa.amplitude_to_db(espectrograma, ref=np.max), sr=sr2, x_axis="time", y_axis="log")
plt.title("Espectrograma")
plt.colorbar(format="%+2.0f dB")
plt.show()
```

```python
fft = np.fft.fft(audio1)  # Transformada de Fourier
frequencies = np.fft.fftfreq(len(fft), 1 / sr1)  # Frequências associadas

# Extrair apenas a parte positiva (útil para análise)
positive_frequencies = frequencies[:len(frequencies) // 2]
magnitude = np.abs(fft[:len(fft) // 2])  # Magnitude das frequências
```

```python
plt.figure(figsize=(10, 6))
plt.plot(positive_frequencies, magnitude)
plt.title("Espectro de Frequências")
plt.xlabel("Frequência (Hz)")
plt.ylabel("Magnitude")
plt.grid()
plt.show()
```
