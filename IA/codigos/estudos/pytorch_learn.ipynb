```python
import torch
import matplotlib.pyplot as plt
```

```python
def func(x):
    return (x - 5) ** 2

def func2(x):
    return x ** 3
```

```python
x_ref = torch.linspace(0, 10, steps=50)
y_ref = func(x_ref)

plt.plot(x_ref, y_ref)
plt.grid(True)
```

```python
x = torch.tensor(2.0)
x.requires_grad_(True)

z = func2(x)
z.backward()
x.grad


```

```python

for _ in range(20):
    y = func(x)
    plt.plot(x.data, y.data, 'ro')
    y.backward()
    x.data -= 0.1 * x.grad
    x.grad.zero_()

plt.plot(x_ref, y_ref)
plt.grid(True)
```
