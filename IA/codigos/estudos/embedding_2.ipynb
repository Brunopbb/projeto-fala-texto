```python
window_size = 3
vocab = set(all_words) #Não mantem a ordem
```

```python
word2idx = {word: idx for idx, word in enumerate(vocab)}
idx2word = {idx : word for idx, word in word2idx.items()}
vocab_size = len(word2idx)
```

```python
word2idx
```

```python
def generate_training_data(corpus, window_size):
    pairs = []

    for idx, center in enumerate(corpus):
        for w in range(-window_size, window_size+1):
            context_idx = idx + w
            if context_idx < 0 or context_idx >= len(corpus) or w == 0:
                continue
            context = corpus[context_idx]
            pairs.append((center, context))
    return pairs

def one_hot(word):
    vec = np.zeros(vocab_size)
    vec[word2idx[word]] = 1
    return vec

def softmax(x):
    e_x = np.exp(x - np.max(x))
    return e_x / e_x.sum()

def cross_entropy(predicted, target):
    
    return -np.log(predicted[target] + 1e-9)
```

```python
pairs = generate_training_data(all_words, window_size)
```

```python
pairs
```

```python
embedding_dim = 100
```

```python
W1 = np.random.randn(vocab_size, embedding_dim) 
W2 = np.random.randn(embedding_dim, vocab_size) 
```

```python
learning_rate = 0.01
epochs = 1000
```

```python
for epoch in range(epochs):
    loss = 0
    for center, context in pairs:
        
        x = one_hot(center)
        
        target = word2idx[context]

        hc = W1[word2idx[center]]

        vc = np.dot(W2.T, hc)

        

        y_pred = softmax(vc)



        loss += cross_entropy(y_pred, target)

        e = y_pred
        e[target] -= 1


        dW2 = np.outer(hc, e)
        dW1 = np.outer(one_hot(center), np.dot(W2, e))

        W1 -= learning_rate * dW1
        W2 -= learning_rate * dW2

    if epoch % 100 == 0:
        print(f"Epoch {epoch}, Loss: {loss:.4f}")


print(f"Epoch {epoch}, Loss: {loss:.4f}")
```
