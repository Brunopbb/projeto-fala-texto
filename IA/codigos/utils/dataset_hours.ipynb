```python
from datasets import load_dataset
import torchaudio
```

```python
def calculate_hours(dataset):

    full_duration = 0.0

    for example in dataset:
        audio = example["audio"]
        full_duration += audio["array"].shape[0] / audio["sampling_rate"]  
    return full_duration 
```

```python
dataset = load_dataset("brunopbb/ufcg-labmet-fala-texto")

train = dataset['train']
test = dataset['test']

duration_test = calculate_hours(test)
duration_train = calculate_hours(train)



min_full_duration = (duration_test + duration_train) / 60

print(f'duração train: {(duration_train / 60):.4f}')
print(f'duração test: {(duration_test / 60):.4f}')
print(f"Duração total do dataset: {min_full_duration:.4f} minutos")
```
