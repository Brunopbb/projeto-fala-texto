```python
from datasets import load_dataset
from huggingface_hub import login
# Carrega o dataset (substitua "nome_do_dataset" pelo nome do dataset que você está usando)
dataset = load_dataset("brunopbb/ufcg-labmet-fala-texto", split="train")  # ou 'test', 'validation', dependendo da divisão
```

```python
lista = ['Punção de cisterna suboccipital', 'Biópsia de fígado em cunha / fragmento']
for i in lista:

  dataset = dataset.filter(lambda example: example['transcription'] != i)
```
